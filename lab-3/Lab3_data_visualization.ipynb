```python
import pandas as pd
import altair as alt
from vega_datasets import data as vega_data
import requests
import io
```

## Load 'Seattle Library Data' and save as data frame

```python
url = "https://raw.githubusercontent.com/manika-lamba/SP26-LIS4_5693/refs/heads/main/lab-assignments/lab-3/Seattle-Library_2015-2021.csv"
response = requests.get(url)
response.raise_for_status() # Raise an exception for HTTP errors
text = response.text
```

```python
# Save data as dataframe
seattle_df = pd.read_csv(io.StringIO(text))
seattle_df.head()
```

## TASK 1: Visualize two columns from data frame

```python
seattle_df['UsageClass'].value_counts()
```

```python
seattle_df['CheckoutType'].value_counts()
```

### Create data frames for the 'UsageClass' and 'CheckoutType' columns

```python
source1 = seattle_df['UsageClass'].value_counts().rename_axis('UsageClass').reset_index(name='counts')
```

```python
source2 = seattle_df['CheckoutType'].value_counts().rename_axis('CheckoutType').reset_index(name='counts')
```

### Create pie chart for the 'UsageClass' column

```python
alt.Chart(source1).mark_arc().encode(
    theta="counts",
    color="UsageClass"
)
```

### Create horizontal bar chart for the 'CheckoutType' column

```python
alt.Chart(source2).mark_bar().encode(
    x="counts",
    y="CheckoutType"
)
```

## TASK 3: Add interactivity

### Adding a click function to the legend the change opacity in the chart

```python
selection = alt.selection_point(fields=['UsageClass'], bind='legend')
alt.Chart(source1).mark_arc().encode(
    theta="counts",
    color="UsageClass",
    opacity=alt.condition(selection, alt.value(1), alt.value(0.2))
).add_params(
    selection
)
```

### Add a hover function the chart that shows the exact count number

```python
alt.Chart(source2).mark_bar().encode(
    x="counts",
    y="CheckoutType",
    tooltip='counts'
).interactive()
```

## TASK 4: Horizontal Concatenation

```python
base1 = alt.Chart(source1).encode(
    theta="counts",
    color="UsageClass",
    opacity=alt.condition(selection, alt.value(1), alt.value(0.2))
).add_params(
    selection
)
base2 = alt.Chart(source2).encode(
    x="counts",
    y="CheckoutType",
    tooltip='counts'
).interactive()
alt.hconcat(base1.mark_arc(),
            base2.mark_bar())
```
